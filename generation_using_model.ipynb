# Generating brain slices with the trained generator

```python
import matplotlib.pyplot as plt
import tensorflow as tf

from models import generator_model, NOISE_DIM
```

```python
checkpoint_dir = "checkpoint_last"
```

```python
generator = generator_model()
```

Only the generator is restored; `expect_partial()` silences the warnings about the optimiser and discriminator values that are in the checkpoint but not needed here.

```python
latest = tf.train.latest_checkpoint(checkpoint_dir)
if latest is None:
    raise FileNotFoundError(f"No checkpoint found in {checkpoint_dir!r}, train the model with 4.GAN.ipynb first")
checkpoint = tf.train.Checkpoint(generator=generator)
checkpoint.restore(latest).expect_partial()
print("Restored", latest)
```

```python
num_images = 9
noise = tf.random.normal([num_images, NOISE_DIM])
generated_images = generator(noise, training=False)

fig = plt.figure(figsize=(10, 10))
for i in range(num_images):
    plt.subplot(3, 3, i + 1)
    plt.imshow((generated_images[i, :, :, 0] + 1) / 2, cmap="gray")
    plt.axis('off')
plt.show()
```
